```python
from pyspark.sql.functions import *
```

```python
source_s3_path = "s3://etl-dataset-ecommerce"
file_format = "csv"  # e.g., json, csv, parquet

# Define Target Bronze Table and Checkpoint
target_bronze_table = "ecomm_etl.bronze.ecomm_etl.bronze"

raw_stream_df_customer = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_customers_dataset.csv')
)
raw_data_df_geolocation = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_geolocation_dataset.csv')
)

raw_data_df_order_items = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_order_items_dataset.csv')
)
raw_data_df_order_payments = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_order_payments_dataset.csv')
)
raw_data_df_order_reviews = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_order_reviews_dataset.csv')
)
raw_data_df_orders = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_orders_dataset.csv')
)
raw_data_df_products = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_products_dataset.csv')
)
raw_data_df_sellers = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/olist_sellers_dataset.csv')
)
raw_data_df_product_category = (
    spark.read.format('csv').option('inferSchema',True)\
                            .option('header',True)\
                            .load('/Volumes/ecomm_etl/raw_s3/ecomm_raw_data_s3/product_category_name_translation.csv')
)
```

```python
# Recreate bronze_df with UC-compatible metadata column
bronze_df_customer = (
    raw_stream_df_customer
    .withColumn("ingestion_timestamp", current_timestamp())
    .withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_geolocation = (
    raw_data_df_geolocation
    .withColumn("ingestion_timestamp", current_timestamp())
    .withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_order_items = (
    raw_data_df_order_items
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_order_payments = (
    raw_data_df_order_payments
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_order_reviews = (
    raw_data_df_order_reviews
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_orders = (
    raw_data_df_orders
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_products = (
    raw_data_df_products
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_sellers = (
    raw_data_df_sellers
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)
bronze_df_product_category = (
    raw_data_df_product_category
    .withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file_name", col("_metadata.file_path"))
)

```

```python
# Write in Bronze schema
bronze_df_customer.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_customer")
bronze_df_geolocation.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_geolocation")
bronze_df_order_items.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_order_items")
bronze_df_order_payments.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_order_payments")
bronze_df_order_reviews.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_order_reviews")
bronze_df_orders.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_orders")
bronze_df_products.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_products")
bronze_df_sellers.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_sellers")
bronze_df_product_category.write.mode("overwrite").saveAsTable("ecomm_etl.bronze.bronze_product_category")
```
