```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.table('ecomm_etl.bronze.bronze_product_category')
```

```python
df.display()
```

```python
df.count()
```

```python
df.filter((col("product_category_name").isNull() | ~col("product_category_name").rlike(r'^[a-zA-Z]'))).display()
```

```python
df = df.withColumn('product_category_name', trim(col('product_category_name')))
```

```python
df = df.withColumn('ingestion_timestamp', current_timestamp())\
       .withColumn('source_file_name',lit('ecomm_etl.bronze.bronze_product_category'))
```

```python
df.write.mode('overwrite').option('overWriteSchema', True).saveAsTable('ecomm_etl.silver.silver_product_category')
```
